```python
import numpy as np

# Create a 1D array
array = np.random.randint(1, 100, 9)
print(array)

# Square root
print(np.sqrt(array))

# Number of dimensions
print(array.ndim)

# Reshape into 3x3 matrix
new_array = array.reshape(3, 3)
print(new_array)

# Check dimensions
print(new_array.ndim)

# Convert 2D array into 1D
print(new_array.ravel())

# Reshape again
newm = new_array.reshape(3, 3)
print(newm)

# Array slicing
print(newm[2, 1:3])
print(newm[1:2, 1:3])
print(new_array[0:3, 0:0])
print(new_array[0:2, 0:1])
print(new_array[0:3, 0:1])
print(new_array[1:3])
```
